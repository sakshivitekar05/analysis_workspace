```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
users = pd.read_csv("data/raw/users.csv")
events = pd.read_csv("data/raw/events.csv")
```

```python
users['signup_date'] = pd.to_datetime(users['signup_date'])

signups_per_day = users.groupby(users['signup_date'].dt.date).size()

signups_per_day.plot()
plt.title("User Signups Per Day")
plt.xlabel("Date")
plt.ylabel("Numbernof Users")
plt.show()

```

```python
print(events.columns)
```

```python
events['timestamp'] = pd.to_datetime(events['timestamp'])

events_per_day = events.groupby(events['timestamp'].dt.date).size()

events_per_day.plot()
plt.title("Events Per Day")
plt.xlabel("Date")
plt.ylabel("Number of events")
plt.show()

```

```python
event_types = events['event_name'].value_counts()

event_types.plot(kind = 'bar')
plt.title("Top Event Types")
plt.xlabel("Event Name")
plt.ylabel("Count")
plt.show()

```
